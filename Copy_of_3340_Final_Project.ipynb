## 💵 Milestone 6: Work Out the Total

**Your job:** Write a function named `order_total` that adds up the prices of every item in the order, then adds sales tax.

**Steps:**
1. Give `tax_rate` a **default value** of `0.08`.
2. Loop over the `order` list.
3. Use `get_price(menu, item)` to get each price. Add it to `subtotal` with `+=`.

**Expected output after the checks:** `PASS` for every line.

````python
# TODO: Write a comment here that says what order_total does.
def order_total(order, menu, tax_rate=0.08):
    subtotal = 0
    for item in order:
        subtotal += get_price(menu, item)
    total = subtotal * (1 + tax_rate)
    return round(total, 2)
````

## 🧾 Milestone 13: Print a Receipt

### Part A: Count the items

**Your job:** Write a function named `count_items` that turns an order list into a dictionary of counts. For example, `['Latte', 'Muffin', 'Latte']` becomes `{'Latte': 2, 'Muffin': 1}`.

**Steps:**
1. Start with an empty dictionary named `counts`.
2. Loop over the order.
3. If the item is already a key, add 1 to its count with `+=`.
4. Otherwise, set its count to `1`.
5. Return `counts`.

### Part B: Print the receipt

**Your job:** Finish `print_receipt`. It uses `count_items` and `order_total`.

**Steps:**
1. Fill in `.items()` to loop over the counts.
2. The line `days, open_hour, close_hour = hours` **unpacks** the tuple into three variables. Fill in the missing name.
3. Fill in the missing parts of the f-strings.

**Expected output for the test call at the bottom:**
```
--- RECEIPT ---
2 x Latte: $9.00
1 x Muffin: $3.25
Total with tax: $13.23
Open Mon-Fri, 7 to 18
```

````python
# TODO: Write a comment here that says what count_items does.
def count_items(order):
    counts = {}
    for item in order:
        if item in counts:
            counts[item] += 1
        else:
            counts[item] = 1
    return counts

# TODO: Write a comment here that says what print_receipt does.
def print_receipt(order, menu):
    print('--- RECEIPT ---')
    counts = count_items(order)
    for item, quantity in counts.items():
        line_total = get_price(menu, item) * quantity
        print(f'{quantity} x {item}: ${line_total:.2f}')
    print(f'Total with tax: ${order_total(order, menu):.2f}')
    days, open_hour, close_hours = hours
    print(f'Open {days}, {open_hour} to {close_hours}')

print_receipt(['Latte', 'Muffin', 'Latte'], menu)
````

## 📊 Milestone 14: The Daily Sales Report

At closing time, the owner wants a report. The `sales_log` list holds every finished order, and each order is itself a list of items.

### Part A: daily_report

**Your job:** Write a function named `daily_report` that returns a dictionary with four numbers about the day.

**Steps:**
1. Loop over `sales_log`. Each `sale` is one order.
2. Add the number of items in the sale to `items_sold` with `len(sale)`.
3. Add the sale's total to `revenue` using `order_total(sale, menu)`.
4. Loop over the items in the sale and add each one to the set `unique`. Use the set method `.add`.
5. Return a dictionary with the keys `'orders'`, `'items'`, `'revenue'`, and `'unique_items'`.

### Part B: print_report

**Your job:** Finish `print_report`. Fill in the last key, `'unique_items'`.

**Expected output for the test call at the bottom:**
```
--- DAILY REPORT ---
Orders: 3
Items sold: 4
Revenue: $15.12
Different items sold: 3
```

````python
# TODO: Write a comment here that says what daily_report does.
def daily_report(sales_log, menu):
    items_sold = 0
    revenue = 0
    unique = set()
    for sale in sales_log:
        items_sold += len(sale)
        revenue += order_total(sale, menu)
        for item in sale:
            unique.add(item)
    return {
        'orders': len(sales_log),
        'items': items_sold,
        'revenue': round(revenue, 2),
        'unique_items': unique
    }

# TODO: Write a comment here that says what print_report does.
def print_report(report):
    print('--- DAILY REPORT ---')
    print('Orders:', report['orders'])
    print('Items sold:', report['items'])
    revenue = report['revenue']
    print(f'Revenue: ${revenue:.2f}')
    print('Different items sold:', len(report['unique_items']))

test_log = [['Latte', 'Muffin'], ['Cookie'], ['Latte']]
print_report(daily_report(test_log, menu))
````

**Check your work:**

````python
check('a day with three orders', daily_report(test_log, menu), {'orders': 3, 'items': 4, 'revenue': 15.12, 'unique_items': {'Latte', 'Muffin', 'Cookie'}})
check('a day with no orders', daily_report([], menu), {'orders': 0, 'items': 0, 'revenue': 0, 'unique_items': set()})
````

````python
# TODO: Write a comment here that says what run_cafe does.
def run_cafe():
    my_order = []
    print(f'Welcome to {cafe_name}!')
    customer = clean_name(input('Your name: '))
    while True:
        print()
        print('1) Menu  2) Menu by category  3) Add item  4) Remove item')
        print('5) View order  6) Allergy-safe items  7) Check out  8) Quit')
        choice = input('Choose 1 to 8: ')
        if choice == '1':
            show_menu(menu)
        elif choice == '2':
            show_menu_by_category(menu, categories)
        elif choice == '3':
            name = clean_name(input('Item name: '))
            if get_price(menu, name) is None:
                print('Sorry, we do not have that item.')
            else:
                quantity = ask_quantity()
                for _ in range(quantity):
                    add_to_order(my_order, menu, name)
                print(f'Added {quantity} x {name}.')
        elif choice == '4':
            name = clean_name(input('Item to remove: '))
            if remove_from_order(my_order, name):
                print(f'Removed one {name}.')
            else:
                print('That item is not in your order.')
        elif choice == '5':
            print('Your order:', my_order)
        elif choice == '6':
            allergy_text = input('Your allergies, separated by commas: ')
            my_allergies = set()
            for word in allergy_text.split(','):
                my_allergies.add(word.strip().lower())
            print('Safe for you:', safe_items(menu, my_allergies))
        elif choice == '7':
            print_receipt(my_order, menu)
            points = add_points(customers, customer, order_total(my_order, menu))
            print(f'{customer} now has {points} points ({get_tier(points)}).')
            sales_log.append(my_order)
            my_order = []
        elif choice == '8':
            print('Goodbye!')
            print_report(daily_report(sales_log, menu))
            break
        else:
            print('Please choose a number from 1 to 8.')
````

### ⭐ Extension A: Discount Codes

Only complete this if you chose A.

**Your job:** Write a function named `apply_discount` that lowers a total when the customer gives a valid code.

**Steps:**
1. `codes` is a dictionary. The key is the code and the value is the discount, like `0.10` for 10 percent.
2. If the code is **in** `codes`, return the total multiplied by `1 - codes[code]`, rounded to 2 decimal places. That line is already written.
3. If the code is not in `codes`, return the total unchanged.

**Expected output after the checks:** `PASS` for every line.

````python
codes = {'SAVE10': 0.10, 'FALL20': 0.20}

# Return the discounted total if the code is valid, otherwise return the unchanged total.
def apply_discount(total, code):
    if code in codes:
        return round(total * (1 - codes[code]), 2)
    else:
        return total
````

### ⭐ Extension B: Best Seller

Only complete this if you chose B.

**Your job:** Write a function named `best_seller` that returns the item that appears most often in an order. If the order is empty, it returns `None`.

**Steps:**
1. Use `count_items(order)` to get the counts dictionary.
2. Loop over `counts.items()`.
3. Keep track of the best item so far. If an item's quantity is greater than `best_count`, it becomes the new best item.
4. Return `best_item`.

**Expected output after the checks:** `PASS` for every line.

````python
# TODO: Write a comment here that says what best_seller does.
def best_seller(order):
    counts = count_items(order)
    best_item = None
    best_count = 0
    for item, quantity in counts.items():
        if quantity > best_count:
            best_item = item
            best_count = quantity
    return best_item
````

### ⭐ Extension C: Tip Options

Only complete this if you chose C.

**Your job:** Write two functions. One adds a tip to a total. The other shows the total for every tip choice.

**Steps:**
1. `tips` is a tuple of tip percentages: `(0.10, 0.15, 0.20)`. It is already written.
2. In `add_tip`, give `tip_percent` a default value of `0.15`.
3. In `tip_choices`, loop over `tips`. For each `percent`, call `add_tip(total, percent)` and add the result to the `results` list.
4. Return `results`.

**Expected output after the checks:** `PASS` for every line.

````python
tips = (0.10, 0.15, 0.20)

# TODO: Write a comment here that says what add_tip does.
def add_tip(total, tip_percent=0.15):
    return round(total * (1 + tip_percent), 2)

# TODO: Write a comment here that says what tip_choices does.
def tip_choices(total):
    results = []
    for percent in tips:
        results.append(add_tip(total, percent))
    return results
````

### ⭐ Extension D: Happy Hour

Only complete this if you chose D.

**Your job:** Write a function named `happy_hour_price` that takes 20 percent off between 2 PM and 4 PM. The hour uses a 24-hour clock, so 14 is 2 PM and 16 is 4 PM. Both 14 and 16 count as happy hour.

**Steps:**
1. Check that `hour` is greater than or equal to 14 **and** less than or equal to 16.
2. If it is, return the price multiplied by `0.8`, rounded to 2 decimal places.
3. Otherwise, return the price unchanged.

**Expected output after the checks:** `PASS` for every line.

````python
# TODO: Write a comment here that says what happy_hour_price does.
def happy_hour_price(price, hour):
    if hour >= 2 and hour <= 16:
        return round(price * 0.8, 2)
    else:
        return price
````

### 🐛 Debugging 1: The Short Total

**Hint:** A function leaves as soon as it reaches `return`. Look at where the `return` line is indented.

````python
# The bug was that the return statement was indented inside the for loop, which made the function return immediately after the first item.
def cart_total(order, menu):
    total = 0
    for item in order:
        total += menu[item]
    return total

print(cart_total(['Latte', 'Muffin'], menu))
````
