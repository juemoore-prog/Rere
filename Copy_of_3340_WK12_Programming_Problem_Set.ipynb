## Part 5: List Comprehensions

### Question 17: Curve the Grades (Loop Version)
The professor adds 5 points to every score.
Build a new list called `curved` using an empty list, a loop, and `append`.

```python
scores = [70, 82, 64, 91]

# Start with an empty list
curved = []

# TODO: loop through scores
# TODO: append each score plus 5 to curved
for s in scores:
    curved.append(s + 5)
print(curved)
```

### Question 18: Curve the Grades (Comprehension Version)
Do the same job as Question 17, but with a list comprehension on a single line.

```python
scores = [70, 82, 64, 91]

# TODO: build curved_2 with a list comprehension
# Pattern: [what to put in  for item in list]
curved_2 = [s + 5 for s in scores]

print(curved_2)
```

### Question 19: Honor Roll
Use a list comprehension with an `if` to make a list of only the scores that are **90 or higher**.

```python
scores = [95, 72, 90, 85, 100, 68]

# TODO: build honor_roll with a list comprehension and an if
# Pattern: [what to put in  for item in list  if test]
honor_roll = [s for s in scores if s >= 90]

print(honor_roll)
```

## Part 6: Putting It All Together

### Question 20: Quiz Report
You are building a small report for a quiz. Complete each step using what you learned.

1. Use a slice to get the **top three** scores from `sorted_scores` (the list is already sorted from highest to lowest).
2. Use a loop to find the **average** of all the scores.
3. Use a list comprehension to make a list of the scores that **passed** (70 or higher).
4. Store the average and the number of passing scores in a **tuple** called `summary`, then print it.

```python
sorted_scores = [98, 91, 85, 77, 70, 64, 52]

# Step 1: slice the top three scores
top_three = sorted_scores[-3:]

# Step 2: add up all the scores with a loop, then divide
total = 0
for score in sorted_scores:
    total += score
# TODO: loop and add each score to total
average = total / len(sorted_scores) # TODO: replace with total divided by the number of scores

# Step 3: comprehension for passing scores
passed = [s for s in sorted_scores if s >= 70]

# Step 4: make a tuple with the average and the number who passed
summary = (average, len(passed))

print(top_three)
print(summary)
```
